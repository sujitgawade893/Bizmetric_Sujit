```python
var = input("What you want..? (Book/Notebook): ").lower()
if var == "book":

    standard = int(input("Enter Standard(1-10): "))

    if 1 <= standard <= 4:
        new_standard = '1st-4th'
    elif 5 <= standard <= 8:
        new_standard = '5th-8th'
    else:
        new_standard = '9th-10th'

    subject = input("Enter Subject (Hindi/Marathi/English/Science/Maths): ")

    my_dict = {
        '1st-4th': {'Books': {'Hindi': 60, 'Marathi': 60, 'English': 80, 'Science': 90, 'Maths': 100}},
        '5th-8th': {'Books': {'Hindi': 100, 'Marathi': 100, 'English': 100, 'Science': 120, 'Maths': 140}},
        '9th-10th': {'Books': {'Hindi': 150, 'Marathi': 150, 'English': 150, 'Science': 200, 'Maths': 250}}
    }

    price = my_dict[new_standard]['Books'][subject]
    print("Book Price:", price)
    quantity = int(input("Enter Quantity: "))
    Final_price=price*quantity
    print("Book Price:", Final_price)
elif var == "notebook":

    pagess = int(input("Enter Pages (100/200): "))
    book_types = input("Enter Book Types (square/4lines/2lines/single lines/A4 notebook): ").lower()

    if pagess == 100:
        new_pages = 'Pages100'
    else:
        new_pages = 'Pages200'

    my_dict_2 = {
        'Pages100': {'square': 40, '4lines': 30, '2lines': 30, 'single lines': 60, 'a4 notebook': 100},
        'Pages200': {'square': 70, '4lines': 50, '2lines': 50, 'single lines': 100, 'a4 notebook': 180}
    }

    price = my_dict_2[new_pages][book_types]
    print("Notebook Price:", price)
    quantity = int(input("Enter Quantity: "))
    Final_price=price*quantity
    print("Notebook Price:", Final_price)
else:
    print("Invalid Choice!")
```
